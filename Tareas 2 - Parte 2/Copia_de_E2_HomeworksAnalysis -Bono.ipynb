# Exercise 2.1

Analyze the writing patterns of each student

```python
import pandas as pd
import numpy as np

# Asumiendo que ya tenemos data cargado
def analizar_patrones_escritura(df):
    # Crear un diccionario para almacenar los resultados por estudiante
    analisis = {
        'longitud_respuestas': [],
        'palabras_por_respuesta': [],
        'genero': []
    }

    # Analizar cada estudiante
    for index, row in df.iterrows():
        # Obtener todas las respuestas excepto Genero
        respuestas = row.drop('Genero').dropna()

        # Calcular métricas
        longitudes = [len(str(r)) for r in respuestas]
        palabras = [len(str(r).split()) for r in respuestas]

        # Guardar resultados
        analisis['longitud_respuestas'].append(np.mean(longitudes))
        analisis['palabras_por_respuesta'].append(np.mean(palabras))
        analisis['genero'].append(row['Genero'])

    # Crear DataFrame con resultados
    resultados = pd.DataFrame(analisis)
    print("\nResumen de patrones de escritura:")
    print(resultados.describe())
    print("\nPromedio por género:")
    print(resultados.groupby('genero').mean())

    return resultados

# Ejecutar el análisis
resultados = analizar_patrones_escritura(data)
```

Del análisis de los patrones de escritura podemos observar que:

En cuanto a la longitud de las respuestas, el promedio general es de 1,826 caracteres, con una desviación estándar considerable de 428 caracteres. La respuesta más corta fue de 643 caracteres, mientras que la más extensa alcanzó 2,685 caracteres. Esto indica una variación significativa en la extensión de las respuestas entre los estudiantes.

Respecto al número de palabras por respuesta, el promedio se sitúa en 282 palabras, con una desviación estándar de 62 palabras. Las respuestas varían desde un mínimo de 107 palabras hasta un máximo de 412 palabras por respuesta.

Al analizar las diferencias por género, se observa que los estudiantes masculinos tienden a escribir respuestas ligeramente más extensas, con un promedio de 1,842 caracteres y 286 palabras por respuesta, en comparación con las estudiantes femeninas, que promedian 1,771 caracteres y 267 palabras por respuesta. Sin embargo, esta diferencia no es dramáticamente significativa.

# Exercise 2.2

Evaluate the similarities of the homeworks of the students.

At a homework level, then as a student level.


```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def analizar_similitudes(df):
    # Inicializamos el vectorizador TF-IDF sin stop_words (podemos preprocesar el texto si es necesario)
    vectorizer = TfidfVectorizer()

    # Analizaremos cada columna de respuestas
    columnas_respuestas = [col for col in df.columns if col != 'Genero']
    resultados = {}

    for columna in columnas_respuestas:
        # Eliminamos valores NaN
        respuestas = df[columna].dropna()

        if len(respuestas) > 1:  # Necesitamos al menos 2 respuestas
            # Convertimos el texto a matriz TF-IDF
            tfidf_matrix = vectorizer.fit_transform(respuestas)

            # Calculamos similitud del coseno
            matriz_similitud = cosine_similarity(tfidf_matrix)

            # Calculamos el promedio de similitud para esta pregunta
            similitud_promedio = np.mean(matriz_similitud[np.triu_indices(matriz_similitud.shape[0], k=1)])

            resultados[columna] = similitud_promedio

    return pd.Series(resultados)

# Ejecutar el análisis
similitudes = analizar_similitudes(data)
print("\nPromedio de similitud por pregunta:")
print(similitudes)
```

Analizando los resultados de similitud entre las respuestas de los estudiantes, podemos observar en cada tema que:

Las respuestas sobre Árboles de Decisión (E6) muestran la mayor similitud con un índice de 0.378, lo que sugiere que los estudiantes tienden a usar terminología y conceptos similares al explicar este tema. Esto podría indicar una comprensión consistente del material o posiblemente el uso de fuentes de información comunes.

Las explicaciones sobre REST-API (E10) y Ensemble Trees (E8) también muestran niveles relativamente altos de similitud, con índices de 0.361 y 0.343 respectivamente. Esto indica que los estudiantes mantienen una consistencia considerable en la forma de describir estos conceptos técnicos.

Por otro lado, el tema de Procesamiento de Lenguaje Natural (E11) muestra la menor similitud con un índice de 0.177, lo que sugiere una mayor diversidad en las respuestas. Esto podría indicar que los estudiantes abordan este tema desde diferentes perspectivas o tienen distintos niveles de comprensión del material.

Los temas de Series de Tiempo (E1) y Datos de Panel (E4) muestran niveles moderados de similitud, con índices de 0.290 y 0.303 respectivamente, lo que sugiere un balance entre respuestas estandarizadas y explicaciones personalizadas.

#### Modelo 2

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder

def crear_clasificador_genero_mejorado(df):
    # Preparar los datos
    X = df.drop('Genero', axis=1).fillna('')
    X_text = X.apply(lambda x: ' '.join(x), axis=1)

    # Codificar las etiquetas
    le = LabelEncoder()
    y = le.fit_transform(df['Genero'])

    # División estratificada
    X_train, X_test, y_train, y_test = train_test_split(
        X_text, y,
        test_size=0.2,
        random_state=42,
        stratify=y
    )

    # Vectorización con más características
    vectorizer = TfidfVectorizer(
        max_features=2000,  # Aumentamos el número de características
        ngram_range=(1, 3),  # Incluimos trigramas
        min_df=2  # Eliminamos términos muy raros
    )
    X_train_vectorized = vectorizer.fit_transform(X_train)
    X_test_vectorized = vectorizer.transform(X_test)

    # Aplicar SMOTE con configuración ajustada
    smote = SMOTE(random_state=42, k_neighbors=3)
    X_train_balanced, y_train_balanced = smote.fit_resample(X_train_vectorized, y_train)

    # Entrenar clasificador con parámetros ajustados
    clf = RandomForestClassifier(
        n_estimators=300,  # Más árboles
        class_weight='balanced_subsample',  # Mejor manejo del desbalance
        max_depth=15,
        min_samples_leaf=2,
        random_state=42
    )
    clf.fit(X_train_balanced, y_train_balanced)

    # Evaluar
    y_pred = clf.predict(X_test_vectorized)
    print("\nReporte de Clasificación:")
    print(classification_report(y_test, y_pred, zero_division=1))

    return clf, vectorizer, X_test, y_test, le

# Ejecutar el clasificador mejorado
modelo_mejorado, vectorizer, X_test, y_test, le = crear_clasificador_genero_mejorado(data)
```

El código se mejoró implementando varias estrategias sofisticadas para manejar el desbalance de clases y mejorar la calidad de la clasificación:

1. Procesamiento de texto mejorado:
- Se aumentó el número de características a 2000 para capturar más información relevante
- Se incluyeron n-gramas hasta trigramas para considerar secuencias de palabras
- Se implementó un filtro min_df para eliminar términos muy raros que podrían ser ruido
1. Manejo del desbalance:
- Se utilizó SMOTE con k_neighbors=3 para generar muestras sintéticas de la clase minoritaria
- Se implementó class_weight='balanced_subsample' en el RandomForest para dar más peso a la clase menos representada
- Se mantuvo la estratificación en la división de datos para preservar la proporción de clases

Analizando los resultados del clasificador:

- Para la clase 0 (femenino), el modelo muestra una precisión perfecta (1.00) pero un recall de 0.00, lo que indica que el modelo es demasiado conservador al predecir esta clase. El f1-score de 0.00 refleja este desequilibrio.
- Para la clase 1 (masculino), el modelo muestra un mejor balance con una precisión de 0.80 y un recall perfecto (1.00), resultando en un f1-score de 0.89. Esto indica que el modelo es muy efectivo identificando estudiantes masculinos.
- La exactitud general (accuracy) es de 0.80, lo que significa que el modelo acierta el 80% de las predicciones totales. El promedio ponderado (weighted avg) de 0.71 en f1-score sugiere un rendimiento moderadamente bueno, aunque todavía hay espacio para mejora, especialmente en la identificación de estudiantes femeninas.

#### Modelo 3

```python
from imblearn.combine import SMOTETomek
from imblearn.ensemble import EasyEnsembleClassifier
from sklearn.metrics import make_scorer, fbeta_score, precision_recall_curve, auc, recall_score, precision_score
from sklearn.model_selection import StratifiedKFold
import numpy as np

def crear_clasificador_genero_avanzado(df):
    # Preparar datos
    X = df.drop('Genero', axis=1).fillna('')
    X_text = X.apply(lambda x: ' '.join(x), axis=1)
    y = df['Genero']

    # Vectorización
    vectorizer = TfidfVectorizer(
        max_features=3000,
        ngram_range=(1, 3),
        min_df=2
    )
    X_vectorized = vectorizer.fit_transform(X_text)

    # Aplicar SMOTETomek para balance de clases
    smt = SMOTETomek(random_state=42)
    X_resampled, y_resampled = smt.fit_resample(X_vectorized, y)

    # Configurar validación cruzada estratificada
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    # Configurar clasificador EasyEnsemble
    classifier = EasyEnsembleClassifier(
        n_estimators=10,
        random_state=42,
        sampling_strategy='auto'
    )

    # Entrenamiento y evaluación con validación cruzada
    scores = []
    for train_idx, val_idx in skf.split(X_resampled, y_resampled):
        X_train, X_val = X_resampled[train_idx], X_resampled[val_idx]
        y_train, y_val = y_resampled[train_idx], y_resampled[val_idx]

        classifier.fit(X_train, y_train)
        y_pred = classifier.predict(X_val)

        # Calcular G-mean
        g_mean = np.sqrt(recall_score(y_val, y_pred, average='weighted') *
                        precision_score(y_val, y_pred, average='weighted'))
        scores.append(g_mean)

    print(f"G-mean promedio en validación cruzada: {np.mean(scores):.3f} ± {np.std(scores):.3f}")

    return classifier, vectorizer

# Ejecutar el clasificador mejorado
modelo_final, vectorizer = crear_clasificador_genero_avanzado(data)
```

El código implementa una solución para manejar el desbalance de clases en nuestro problema de clasificación de género.

1. Preprocesamiento de texto y vectorización:
- Usamos TfidfVectorizer con 3000 características y n-gramas hasta 3 palabras para capturar mejor el contexto y patrones en las respuestas
- El parámetro min_df=2 elimina términos que aparecen en un solo documento, reduciendo el ruido
1. Manejo del desbalance de clases:
- SMOTETomek: Combina la generación de muestras sintéticas (SMOTE) con la eliminación de ejemplos ruidosos (Tomek links)
- Esta técnica es más sofisticada que SMOTE simple porque no solo crea nuevos ejemplos de la clase minoritaria, sino que también limpia los datos eliminando pares de ejemplos que podrían estar causando confusión en el límite de decisión
1. Clasificación con EasyEnsemble:
- Utiliza múltiples clasificadores (n_estimators=10) que se entrenan en diferentes submuestras balanceadas de los datos
- sampling_strategy='auto' ajusta automáticamente la proporción de clases para un balance óptimo
1. Validación y evaluación:
- Implementamos validación cruzada estratificada con 5 folds para mantener la proporción de clases en cada partición
- Usamos G-mean como métrica principal porque es más apropiada para datos desbalanceados que la accuracy, ya que considera tanto la precisión como el recall

Esta implementación busca resolver el problema del desbalance extremo (36 hombres vs 10 mujeres) mediante una combinación de técnicas de muestreo avanzadas y un ensamble de clasificadores, mientras mantiene una evaluación robusta del rendimiento.

El G-mean de 0.851 ± 0.070 representa una mejora significativa sobre nuestros intentos anteriores. Analicemos qué significa:

1. El valor de 0.851 indica que el modelo está logrando un buen balance entre precisión y recall para ambas clases (masculino y femenino). Es mucho mejor que nuestros resultados anteriores donde teníamos problemas identificando la clase minoritaria (femenino).
1. La desviación estándar de 0.070 sugiere que el modelo es bastante estable, con variaciones relativamente pequeñas entre diferentes folds de la validación cruzada.
1. Este resultado demuestra que nuestra estrategia combinada de:
- SMOTETomek para el balance de clases
- EasyEnsemble para la clasificación
- Validación cruzada estratificada

ha sido efectiva para manejar el desbalance de clases y mejorar la capacidad del modelo para identificar correctamente ambos géneros.

# Exercise 2.4
Classify group members into different subgroups (minimum 3) according to a characteristic of your choice.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

def clasificar_subgrupos(df, n_clusters=3):
    # Preparar características para clustering
    # 1. Longitud promedio de respuestas
    longitud_respuestas = df.drop('Genero', axis=1).apply(
        lambda x: x.apply(lambda y: len(str(y)) if pd.notna(y) else 0)
    ).mean(axis=1)

    # 2. Número de respuestas completas
    respuestas_completas = df.drop('Genero', axis=1).notna().sum(axis=1)

    # Crear DataFrame de características
    features = pd.DataFrame({
        'longitud_promedio': longitud_respuestas,
        'respuestas_completas': respuestas_completas
    })

    # Normalizar características
    scaler = StandardScaler()
    features_scaled = scaler.fit_transform(features)

    # Aplicar K-means
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    clusters = kmeans.fit_predict(features_scaled)

    # Agregar clusters al DataFrame original
    df_con_clusters = df.copy()
    df_con_clusters['Cluster'] = clusters

    # Mostrar estadísticas por cluster
    print("\nEstadísticas por cluster:")
    for i in range(n_clusters):
        print(f"\nCluster {i}:")
        print(f"Número de estudiantes: {sum(clusters == i)}")
        print(f"Longitud promedio de respuestas: {longitud_respuestas[clusters == i].mean():.2f}")
        print(f"Promedio de respuestas completas: {respuestas_completas[clusters == i].mean():.2f}")

    return df_con_clusters, kmeans

# Ejecutar la clasificación
df_clusters, modelo = clasificar_subgrupos(data)

```

El código implementa un algoritmo de clasificación no supervisada (K-means) para agrupar a los estudiantes en tres subgrupos basándose en dos características principales:

1. Longitud promedio de respuestas: Mide cuánto escriben los estudiantes en promedio
1. Número de respuestas completas: Indica cuántas preguntas respondió cada estudiante
